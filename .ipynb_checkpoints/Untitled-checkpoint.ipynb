```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel("HR_Attrition_Dataset.xlsx")
df.info()
```

```python
df.isnull().sum()
df["Department"]=df["Department"].str.strip().str.title()
```

```python
df=df.drop_duplicates()
df.duplicated().sum()
```

```python
# grouping attrition based on department
department = df.groupby("Department").agg(
    Total=('EmployeeID', 'count'),
    Left=('Attrition', lambda x: (x == 'Yes').sum())
)
department['Attrition_pct'] = (department['Left'] / department['Total'] * 100).round(2)
print(department)
```

```python
sns.barplot(x=department.index,y=department["Attrition_pct"])
plt.xticks(rotation=45)
plt.show()
```

```python
# grouping attrition based on gender
gender_grp=df.groupby("Gender").agg(
    Total=("EmployeeID","count"),
    Left=('Attrition', lambda x:(x=="Yes").sum())
)
gender_grp["Left%"]=(gender_grp["Left"]/gender_grp["Total"]*100).round(2)
print(gender_grp)
```

```python
# grouping attrition based on location
location_grp=df.groupby("Location").agg(
      Total=("EmployeeID","count"),
    Left=('Attrition', lambda x:(x=="Yes").sum())
)
location_grp["Left%"]=(location_grp["Left"]/location_grp["Total"]*100).round(2)
print(location_grp)
```

```python
#  Grouping Attriton based on Overtime
overtime_grp=df.groupby("OverTime").agg(
      Total=("EmployeeID","count"),
    Left=('Attrition', lambda x:(x=="Yes").sum())
)
overtime_grp["Left%"]=(overtime_grp["Left"]/overtime_grp["Total"]*100).round(2)
print(overtime_grp)
```

```python
plt.pie(
    x=overtime_grp["Left%"],
    labels=overtime_grp.index,
    autopct='%1.1f%%'
)
plt.show()
```

```python
# attrition based on Age
df["Age_Band"]=pd.cut(df["Age"],bins=[18,25,35,45,55,65],labels=["18-25","26-35","36-45","46-55","56-65"])
Age_Band_grp=df.groupby("Age_Band",observed=True).agg(
    Total=("EmployeeID","count"),
    Left=("Attrition", lambda x: (x=="Yes").sum()),
    Left_Male=("Gender", lambda x: ((x == "Male") & (df.loc[x.index, "Attrition"] == "Yes")).sum()),
    Left_Female=("Gender", lambda x: ((x == "Female") & (df.loc[x.index, "Attrition"] == "Yes")).sum()),
    Left_Other=("Gender", lambda x: ((x == "Other") & (df.loc[x.index, "Attrition"] == "Yes")).sum())
)

Age_Band_grp["Attrition%"]=(Age_Band_grp["Left"]/Age_Band_grp["Total"]*100).round(2)
print(Age_Band_grp)
sns.barplot(x=Age_Band_grp.index,y=Age_Band_grp["Attrition%"])
plt.show()
```

```python
# attrition based on hike
df["hike_band"] = pd.cut(
    df["LastSalaryHikePct"],
    bins=[-1, 4, 8, 12, 16, 20],
    labels=["0-4", "5-8", "9-12", "13-16", "17-20"]
)

hike=df.groupby("hike_band",observed=True).agg(
    Total=("EmployeeID","count"),
    Left=("Attrition",lambda x:(x=="Yes").sum())
)
hike["Left%"]=(hike["Left"]/hike["Total"]*100).round(2)
print(hike)
```

```python
# attrition based on workmode
workMode_grp=df.groupby("WorkMode").agg(
            Total=("EmployeeID","count"),
            Left=("Attrition",lambda x:(x=="Yes").sum())
)
workMode_grp["Left%"]=(workMode_grp["Left"]/workMode_grp["Total"]*100).round(2)
print(workMode_grp)
```

```python
#  promotion group
promotion_grp=df.groupby("YearsSinceLastPromotion").agg(
            Total=("EmployeeID","count"),
            Left=("Attrition",lambda x:(x=="Yes").sum())
)
promotion_grp["Left%"]=(promotion_grp["Left"]/promotion_grp["Total"]*100).round(2)
print(promotion_grp)
```

```python
# job satifaction group
job_satisfaction_grp=df.groupby("JobSatisfaction").agg(
            Total=("EmployeeID","count"),
            Left=("Attrition",lambda x:(x=="Yes").sum())
)
job_satisfaction_grp["Left%"]=(job_satisfaction_grp["Left"]/job_satisfaction_grp["Total"]*100).round(2)
print(job_satisfaction_grp)
```

```python
# work_life_balance_grp
work_life_balance_grp=df.groupby("WorkLifeBalance").agg(
            Total=("EmployeeID","count"),
            Left=("Attrition",lambda x:(x=="Yes").sum())
)
work_life_balance_grp["Left%"]=(work_life_balance_grp["Left"]/work_life_balance_grp["Total"]*100).round(2)
print(work_life_balance_grp)
```

```python
# Salary gap 
level_gap = df.pivot_table(
    index="JobLevel",
    columns="Department",
    values="MonthlySalary",
    aggfunc="median"
)
print(level_gap)
```

```python
# Salary gap based on gender
level_gap = df.pivot_table(
    index="JobLevel",
    columns="Gender",
    values="MonthlySalary",
    aggfunc="median"
)
level_gap["Gap%"] = ((level_gap["Female"] - level_gap["Male"]) / level_gap["Male"] * 100).round(2)
print(level_gap)
```

```python
# reason based on department
left = df[df["Attrition"] == "Yes"]

print(left["ExitReason"].value_counts())
print(pd.crosstab(left["ExitReason"], left["Department"]))
print(pd.crosstab(left["ExitReason"], left["Department"], normalize="columns").mul(100).round(1))

pd.crosstab(left["ExitReason"], left["Department"]).plot(kind="bar")
plt.show()
```

```python
# employees who left within 2 years of joining
left = df[df["Attrition"] == "Yes"]
earlyleft = left[(left["ExitDate"].dt.year - left["JoinDate"].dt.year) <= 2]

print("Early leavers %:", round(len(earlyleft) / len(left) * 100, 1))
```

```python
df["Peer"]=df.groupby(["JobLevel","Department"])["MonthlySalary"].transform("median")
df["CompRatio"]=(df["MonthlySalary"]/df['Peer']).round(2)
df['Tenure']=((pd.Timestamp.today()-df["JoinDate"]).dt.days/365).round()

df["RiskScore"] = 0

df.loc[df["CompRatio"] < 0.9, "RiskScore"] += 2            # paid below peers
df.loc[df["Tenure"] <= 2, "RiskScore"] += 2                 # new employee
df.loc[df["Age"] <= 25, "RiskScore"] += 1                   # young age band
df.loc[df["LastSalaryHikePct"] <= 4, "RiskScore"] += 1      

active=df[df["Attrition"]=="No"]
top_risk=active.sort_values("RiskScore",ascending=False).head(10)
print(top_risk[["EmployeeID","Department","JobLevel","Tenure","RiskScore","CompRatio","Peer"]].to_string(index=False))
```
